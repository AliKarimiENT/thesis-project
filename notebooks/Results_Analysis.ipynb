# 📊 QG Experiment - Results Analysis

Comprehensive analysis of difficulty-controlled question generation results.

```
import sys
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from pathlib import Path

sys.path.append(str(Path.cwd().parent))

from src.utils.io_utils import load_jsonl, load_json

print('✅ Imports successful!')
```

```
# Load results
baseline_preds = load_jsonl('../reports/tables/preds_baseline_squad.jsonl')
controlled_preds = load_jsonl('../reports/tables/preds_controlled_squad.jsonl')
baseline_metrics = load_json('../reports/tables/metrics_baseline_squad.json')
controlled_metrics = load_json('../reports/tables/metrics_controlled_squad.json')

print(f'Loaded {len(baseline_preds)} predictions')
```

```
# Metrics comparison
comparison = pd.DataFrame({
    'Metric': ['BLEU', 'ROUGE-1', 'ROUGE-2', 'ROUGE-L'],
    'Baseline': [baseline_metrics['bleu'], baseline_metrics['rouge1'], baseline_metrics['rouge2'], baseline_metrics['rougeL']],
    'Controlled': [controlled_metrics['bleu'], controlled_metrics['rouge1'], controlled_metrics['rouge2'], controlled_metrics['rougeL']]
})
print(comparison)
```

```
# Visualize
import matplotlib.pyplot as plt

comparison.set_index('Metric')[['Baseline', 'Controlled']].plot(kind='bar', figsize=(10,6))
plt.title('Metrics Comparison')
plt.ylabel('Score')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
